```python
from pathlib import Path 
import json
file_1 = Path('purchase_log.txt')
file_2 = Path('visit_log.csv')
purchases = {}

with open(file_1, 'r', encoding='utf-8') as file:
    for line in file:
        purchase = json.loads(line)
        user_id = purchase['user_id']
        category = purchase['category']
        purchases[user_id] = category

with open(file_2, 'r', encoding='utf-8') as visits:
    with open('funnel.csv', 'w', encoding='utf-8') as funnel:
        funnel.write('user_id,source,category\n')
        next(visits)

        for line in visits:
            user_id, source = line.strip().split(',')

            if user_id in purchases:
                category=purchases[user_id]
                funnel.write(f'{user_id},{source},{category}\n')
```
